# 7. Sum of weights

```python
df.head()
```

```python
# Select all the cars from Asia
new_df = df[df['origin'] == 'Asia']

new_df.head()
```

```python
# Select only columns vehicle_weight and model_year
asia_df = new_df[['vehicle_weight', 'model_year']]

asia_df.head()
```

```python
# Select the first 7 values
asia_df_new = asia_df.head(7)

asia_df_new
```

```python
# Get the underlying NumPy array
X = asia_df_new.values

X
```

```python
# Compute X transpose multiplied by X
XTX = X.T @ X

XTX
```

```python
# Invert XTX
XTX_inv = np.linalg.inv(XTX)

XTX_inv
```

```python
# Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y
```

```python
# Calculate w
w = XTX_inv @ X.T @ y

w
```

```python
# Sum all elements of w
w.sum().round(3)
```

### Sum of all weights is 0.369
